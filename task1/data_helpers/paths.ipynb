```python
import os
os.mkdir('../finetuned_models_new')

os.listdir('../finetuned_models_new')
```

```python
epochs = [1,2,3,4]
r_alpha = [16,32,64,128]

for epoch in epochs:
    for alpha in r_alpha:
        if not os.path.exists(f"../finetuned_models_new/finetuned_models_new_{epoch}_{alpha}"):
            os.makedirs(f"../finetuned_models_new/finetuned_models_new_{epoch}_{alpha}")
```
